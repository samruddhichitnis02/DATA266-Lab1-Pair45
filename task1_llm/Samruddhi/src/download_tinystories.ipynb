```python
import sys
import torch

from pathlib import Path
from datasets import load_dataset
```

```python
# Create the shared Task 1 data folder
data_dir = Path.cwd() / "task1_llm" / "data"
data_dir.mkdir(parents=True, exist_ok=True)

train_file = data_dir / "tinystories_train_100k.txt"
val_file = data_dir / "tinystories_val_10k.txt"

# Stream the dataset so we do not download the entire TinyStories dataset
dataset = load_dataset(
    "roneneldan/TinyStories",
    split="train",
    streaming=True
)

with open(train_file, "w", encoding="utf-8") as train_output, \
     open(val_file, "w", encoding="utf-8") as val_output:

    for index, example in enumerate(dataset):
        if index >= 110_000:
            break

        text = example["text"].strip().replace("\n", " ")

        if index < 100_000:
            train_output.write(text + "\n")
        else:
            val_output.write(text + "\n")

        if (index + 1) % 10_000 == 0:
            print(f"Processed {index + 1:,} stories")

print("Dataset preparation completed.")
print(f"Training file: {train_file}")
print(f"Validation file: {val_file}")
```
